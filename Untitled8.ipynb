```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load model
model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Long text
text = """
Artificial Intelligence (AI) has become one of the most important technologies
of the modern era and is transforming the way people live, work, communicate,
and solve problems. AI refers to the ability of computer systems to perform
tasks that normally require human intelligence, such as understanding language,
recognizing images, making decisions, learning from data, and predicting future
outcomes. Businesses are using AI to automate repetitive tasks, improve customer
service, analyze large amounts of information, detect fraud, and make better
business decisions. In healthcare, AI can assist doctors by analyzing medical
images, identifying patterns in patient data, supporting disease detection, and
helping researchers discover new medicines. In education, AI-powered applications
can provide personalized learning experiences, automatically evaluate certain
types of assignments, generate educational content, and help students understand
difficult concepts.

Natural Language Processing, also known as NLP, allows computers to understand
and process human language. Applications such as chatbots, language translation,
speech recognition, sentiment analysis, and text summarization depend on NLP
technologies. Text summarization is particularly useful because people and
organizations generate enormous amounts of information every day through
articles, reports, research papers, emails, websites, and social media. Reading
every document completely can require a significant amount of time. Automatic
summarization systems can identify important information and produce a shorter
version of a long document while attempting to preserve its main ideas.

Modern summarization systems often use deep learning models that can understand
relationships between words and sentences. These systems can help students,
researchers, journalists, businesses, and professionals quickly understand
large amounts of information. However, artificial intelligence also introduces
challenges related to privacy, security, bias, misinformation, transparency,
employment, and responsible use. AI systems learn from data, and if the training
data contains errors or biases, the resulting system may produce inaccurate or
unfair outcomes.

As AI continues to develop, it is likely to become increasingly integrated into
everyday applications and professional environments. The future of AI will
depend not only on technological progress but also on how responsibly society
develops, deploys, and regulates these systems. By combining human creativity,
critical thinking, and decision-making with the analytical capabilities of AI,
organizations and individuals can use intelligent technologies to improve
productivity, solve complex problems, and create new opportunities.
"""

# Convert text into tokens
inputs = tokenizer(
    text,
    return_tensors="pt",
    max_length=1024,
    truncation=True
)

# Generate summary
summary_ids = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_length=150,
    min_length=40,
    num_beams=4,
    length_penalty=2.0,
    early_stopping=True
)

# Convert tokens back to text
summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("========== TEXT SUMMARY ==========\n")
print(summary)
```
